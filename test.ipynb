```python
import time
import json

results = []

for i, article in enumerate(articles):
    print(f"[{i+1}/{len(articles)}] {article['id']}: {article['title'][:60]}...")
    start = time.time()
    try:
        tags = tag_article(article, companies, profile)
        elapsed = time.time() - start
        print(f"    -> {len(tags)} tags in {elapsed:.1f}s")
        results.append({
            "article_id": article["id"],
            "title": article["title"],
            "elapsed_seconds": round(elapsed, 1),
            "tags": tags,
            "error": None
        })
    except Exception as e:
        elapsed = time.time() - start
        print(f"    -> FAILED after {elapsed:.1f}s: {e}")
        results.append({
            "article_id": article["id"],
            "title": article["title"],
            "elapsed_seconds": round(elapsed, 1),
            "tags": [],
            "error": str(e)
        })

# Save so you don't lose it
with open("batch_results.json", "w") as f:
    json.dump(results, f, indent=2)

# Quick summary
total_time = sum(r["elapsed_seconds"] for r in results)
failures = [r for r in results if r["error"]]
print(f"\nDone. Total time: {total_time:.1f}s ({total_time/60:.1f} min)")
print(f"Failures: {len(failures)}/{len(results)}")
for r in failures:
    print(f"  - {r['article_id']}: {r['error']}")
```
